# Data Preprocessing

# creating a pipeline for the datapreprocessing 

```python
# we basically create 4 pipline in the data preprocessing as folllows 
#1) skewness_feature pipeline-- In this we do log transformation to make the data more normal distribution and rthen standard scaling
#2)other_numeric_feature_pipeline = In this we do the standard scaling to make the data more normal distribution
#3)ordinary_enccoding_feature_pipeline =it just like label enconging which we do for the odrederd column  of the category 
#4) one_hot_encoding_feature_piepline = we do the one hot encoding for the nominal data type of the category column
```

```python
# Now we will do the data seperation or you can say feature seperation basewd on the skewness , and the encoding type 
# here Time_of_Booking  is the sequesnce or order base so we will do the ordinary encoding and the Vehicle_Type is the nominal data so we will do the one hot encoding for that

skewness_feature=["Number_of_Drivers"]
other_numeric_feature=["Number_of_Riders","Number_of_Past_Rides","Expected_Ride_Duration"]
ordinary_encoding_feature=["Time_of_Booking"]
one_hot_encoding_feature=["Vehicle_Type"]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,FunctionTransformer,OrdinalEncoder,OneHotEncoder

# 1)Skewness_feature pipeline
skewness_feature_pipeline=Pipeline(steps=[
    ("log_transform",FunctionTransformer(np.log1p)),
    ("scaler",StandardScaler())
])

#2)other numeric feature pipeline
other_numeric_feature_pipeline=Pipeline(steps=[
    ("scaler",StandardScaler())
])

#3)ordinary encoding feature pipeline
ordinary_encoding_feature_pipeline=Pipeline(steps=[
    ("encoder",OrdinalEncoder(categories=[["Morning","Afternoon","Evening","Night"]],))
])

#4)one hot encoding feature pipeline
one_hot_encoding_feature_pipeline=Pipeline(steps=[
    ("encoder",OneHotEncoder(handle_unknown="ignore"))
])


# Noe we w
```

```python
# hERE WE HAVE BUILT SIMPLE MINI ENCODING PIPELINE BUT IYTS NOT FINAL ONE 
```

```python
# TRAIN TEST Split
from sklearn.model_selection import train_test_split
x=data.drop("Historical_Cost_of_Ride",axis=1)
y=data["Historical_Cost_of_Ride"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
x_train
```

# now we willl build pipeline for the Encoding (preprocessing)


```python
#ill do the column transformer to combine all the above 4 pipeline into one single pipeline 

from sklearn.compose import ColumnTransformer


ALL_PROCESSOR=ColumnTransformer(transformers=[
    ("skewness_feature_pipeline",skewness_feature_pipeline,skewness_feature),
    ("other_numeric_feature_pipeline",other_numeric_feature_pipeline,other_numeric_feature),
    ("ordinary_encoding_feature_pipeline",ordinary_encoding_feature_pipeline,ordinary_encoding_feature),
    ("one_hot_encoding_feature_pipeline",one_hot_encoding_feature_pipeline,one_hot_encoding_feature)
])

```

# Bulding the final and Imopoetant pipeline which (preprocessing (Encoding).   +.   Model Traing )

```python
# Model training and Encoding pipeline 
from sklearn.linear_model import LinearRegression
linear_pipeline=Pipeline(steps=[
    ("preprocessing",ALL_PROCESSOR),
    ("model",LinearRegression())
])
```

```python
# Training the model via pipeline with X_training and x_test data
linear_pipeline.fit(x_train,y_train)
```

# prediction of the data

```python
# Prdicting the data which the y_train data 
prediction_data=linear_pipeline.predict(x_test)
```

# model Evaluation 

```python
# for model evluation we will use the mean squared error and the r2 score to check the model performance, Root means square Error as well 
from sklearn.metrics import r2_score,mean_absolute_error
r2=r2_score(y_test,prediction_data)
print("R2 score of the model is : ",r2)

# calculate the rmse and the mean Absolute Error 
MAE=mean_absolute_error(y_test,prediction_data)
print("Mean Absolute Error of the model is : ",MAE)

```
